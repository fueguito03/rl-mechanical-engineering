# Results

Increasing the action space from 3 actions to 11 actions did not change the ranking of the epsilon strategies.

The performance ranking was:

1. `constant_01`
2. `decay`
3. `sudden_drop`
4. `constant_1`

Therefore, the previous hypothesis that `constant_01` performed best mainly because the action space was small was not supported by this experiment.

With 11 actions, `constant_01` showed weaker performance during the early episodes. This may be because the agent had more possible actions to explore.

`Sudden_drop` showed the largest increase in reward after epsilon suddenly decreased. However, a larger increase does not necessarily mean better final performance because its initial reward was also lower.

Both `constant_01` and `decay` achieved episode rewards significantly above 100 with 11 actions. This may have resulted from the finer control provided by the larger action space rather than better exploration alone.

Overall, a constant epsilon of 0.1 remained the best strategy in this environment. However, experiments with multiple random seeds and additional trajectory metrics would be necessary to make a stronger conclusion.

# Results

The constant epsilon strategy with ε = 0.1 achieved the highest success rate and learned the fastest.

However, the CarEnv has only three possible actions: `[-1, 0, 1]`. Because the action space is small, even a low exploration rate may have been sufficient to explore the available actions. Therefore, high exploration rates may not have provided much additional benefit in this environment.

In a future experiment, I will test whether increasing the number of possible actions changes the effect of different epsilon strategies.
